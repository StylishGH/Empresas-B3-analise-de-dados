```python
import sqlite3

import pandas as pd

df_roe= pd.read_csv("../data/processed/roe_2021_2025.csv")

sql = sqlite3.connect("../data/processed/empresa_b3.db")
df_roe.to_sql("roe", sql, if_exists="replace", index=False)



```

```python
teste = pd.read_sql("Select * from roe", sql)
teste
#SQL Checkp
```

```python
query = """
SELECT 
    DENOM_CIA,
    AVG(ROE) as ROE_MEDIO,
    COUNT(*) as QTD_ANOS
FROM roe
WHERE ROE_ENGANOSO = 0
GROUP BY DENOM_CIA
HAVING COUNT(*) >= 3
ORDER BY ROE_MEDIO DESC
LIMIT 10
"""
# Ou poderiamos escrever assim: query = "SELECT DENOM_CIA, AVG(ROE) as ROE_MEDIO, COUNT(*) as QTD_ANOS FROM roe WHERE ROE_ENGANOSO = 0 GROUP BY DENOM_CIA HAVING COUNT(*) >= 3 ORDER BY ROE_MEDIO DESC LIMIT 10"

top10_sql = pd.read_sql(query, sql)
top10_sql
```

```python
Top_lucros = """
SELECT
    DENOM_CIA,
    SUM(VL_CONTA_REAIS) as LUCRO_TOTAL
FROM roe
GROUP BY DENOM_CIA
ORDER BY LUCRO_TOTAL DESC
LIMIT 10
"""

top10_lucros = pd.read_sql(Top_lucros, sql)
top10_lucros
```
