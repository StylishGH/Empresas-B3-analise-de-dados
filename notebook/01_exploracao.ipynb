```python
import pandas as pd

anos = [2021, 2022, 2023, 2024, 2025]
dfs = []

for ano in anos:
    caminho = f"../data/raw/DRE/dfp_cia_aberta_DRE_con_{ano}.csv"
    df = pd.read_csv(caminho, sep=";", encoding="latin1")
    df["ANO_ARQUIVO"] = ano
    dfs.append(df)
    print(f"{ano}: {df.shape[0]} linhas")  # confere se algum ano veio vazio/suspeito

df_dre = pd.concat(dfs, ignore_index=True)
print("Total:", df_dre.shape)
df_dre.head()
```

```python
caminho_tipo = "../data/raw/TIPO/cad_cia_aberta.csv"
df_tipo = pd.read_csv(caminho_tipo,sep=";",encoding="latin1")
df_tipo.columns.to_list() # procurando o Setor da empresa
```

```python
df_lucro = df_dre[df_dre["CD_CONTA"] == "3.11"].copy()
df_lucro.head()
```

```python
# antes de somar qualquer coisa, vou ver como o arquivo vem
# porque desconfiei que tem linha repetida pro mesmo ano
print(df_lucro["ORDEM_EXERC"].value_counts())
print("maior versão por empresa/ano:", df_lucro.groupby(["DENOM_CIA", "ANO_ARQUIVO"])["VERSAO"].max().max())
df_lucro[df_lucro["DENOM_CIA"] == "PETROLEO BRASILEIRO S.A. PETROBRAS"][["ANO_ARQUIVO", "DT_REFER", "VERSAO", "ORDEM_EXERC", "VL_CONTA"]].sort_values(["ANO_ARQUIVO", "VERSAO"]).head(12)
```

```python
# achei o problema: o arquivo traz o ano atual (ÚLTIMO) e o ano anterior (PENÚLTIMO)
# como comparativo, e ainda tem empresa que reapresentou o balanço (mais de uma VERSAO)
# se somar tudo, o lucro sai duplicado. então:
# 1. fico só com o ÚLTIMO (o ano de verdade do arquivo)
# 2. fico só com a maior VERSAO de cada empresa/ano
df_lucro = df_lucro[df_lucro["ORDEM_EXERC"] == "ÚLTIMO"].copy()
df_lucro = df_lucro.sort_values("VERSAO").drop_duplicates(subset=["DENOM_CIA", "ANO_ARQUIVO"], keep="last").copy()
print("linhas depois do filtro:", len(df_lucro))
print("duplicados restantes (empresa/ano):", df_lucro.duplicated(subset=["DENOM_CIA", "ANO_ARQUIVO"]).sum())
```

```python
# outra: nem todo mundo reporta em milhares. a Vivara e a Metisa, por exemplo,
# reportam em unidade. pra somar certo, padronizo tudo em milhares antes de agrupar
print(df_lucro["ESCALA_MOEDA"].value_counts())
escala = df_lucro["ESCALA_MOEDA"].astype(str).str.upper()
df_lucro["VL_CONTA_MIL"] = df_lucro["VL_CONTA"]
df_lucro.loc[~escala.str.contains("MIL", na=False), "VL_CONTA_MIL"] = df_lucro["VL_CONTA"] / 1000
```

```python
df_lucro_ano = df_lucro.groupby(["DENOM_CIA", "SETOR_ATIV", "ANO_ARQUIVO"])["VL_CONTA_MIL"].sum().reset_index()
df_lucro_ano = df_lucro_ano.rename(columns={"VL_CONTA_MIL": "VL_CONTA"})
#quero ver o lucro por ano, por empresa e por setor.
df_lucro_ano = df_lucro_ano.sort_values(["ANO_ARQUIVO", "SETOR_ATIV", "VL_CONTA"], ascending=[True, True, False])
#quero fazer em ordem decrescente, do maior lucro para o menor lucro, por ano e por setor.
df_lucro_ano.head(20)
```
